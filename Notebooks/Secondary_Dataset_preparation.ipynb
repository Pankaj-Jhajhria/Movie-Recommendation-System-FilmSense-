```python
import pandas as pd
import re

def combine_features(df, feature_weights, new_col_name='combined_features', sep=' '):
  
    df = df.copy()

    missing = [f for f in feature_weights if f not in df.columns]
    if missing:
        raise ValueError(f"These columns are not in the dataframe: {missing}")

    def tokenize(value):
        if pd.isna(value):
            return []
        value = str(value)
        tokens = re.split(r'[^a-zA-Z]+', value.strip())
        return [t.strip() for t in tokens if t.strip() != '']

    def process_row(row):
        words = []
        for feature, weight in feature_weights.items():
            tokens = tokenize(row[feature])
            words.extend(tokens * weight)  # repeat tokens `weight` times
        return sep.join(words)

    df[new_col_name] = df.apply(process_row, axis=1)
    return df



```

```python
feature_weights = {
    'genres_list': 3,
    'keywords': 3,
    'Director': 3,
    'Star1': 2,
    'Star2': 2,
    'Star3': 2,
    'Star4': 2,
    'Producers': 1,
    'production_companies': 1,
    'overview': 1
}

df = combine_features(df, feature_weights, new_col_name='tags')
df.head(3)
```
