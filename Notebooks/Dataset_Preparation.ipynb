```python
import pandas as pd
import re

def combine_features(df, feature_list, new_col_name='combined_features', sep=' '):
    """
    Combines multiple string-based columns of a dataframe into a single new column.
    Any character that is NOT an alphabet letter (a-z, A-Z) is treated as a separator
    (this includes commas, spaces, periods, quotes, digits, hyphens, colons, etc.).

    Parameters:
        df (pd.DataFrame): The input dataframe.
        feature_list (list of str): List of column names to combine.
        new_col_name (str): Name of the new combined feature column.
        sep (str): Separator to join the final combined words with (default: single space).

    Returns:
        pd.DataFrame: The dataframe with the new combined feature column added.
    """
    df = df.copy()

    # Validate that all requested features exist in the dataframe
    missing = [f for f in feature_list if f not in df.columns]
    if missing:
        raise ValueError(f"These columns are not in the dataframe: {missing}")

    def process_row(row):
        words = []
        for feature in feature_list:
            value = row[feature]

            # Handle NaN / None gracefully
            if pd.isna(value):
                continue

            value = str(value)

            # Split on ANY non-alphabet character (letters only survive as tokens)
            tokens = re.split(r'[^a-zA-Z]+', value.strip())
            tokens = [t.strip() for t in tokens if t.strip() != '']

            words.extend(tokens)

        return sep.join(words)

    df[new_col_name] = df.apply(process_row, axis=1)
    return df
```

```python
import pandas as pd
import numpy as np

def smart_merge(df1, df2, key='title', priority='df1'):
    """
    Merges two dataframes on `key`, combining duplicate rows by filling
    nulls from one another instead of discarding data. When both sides
    have a value, the `priority` dataset's value wins.

    Parameters:
        df1, df2 (pd.DataFrame): Dataframes to merge.
        key (str): Column used to identify duplicate records (e.g. 'title').
        priority (str): 'df1' or 'df2' — which dataset's non-null value 
                         wins when both have a value for the same column.

    Returns:
        pd.DataFrame: One row per unique key, with maximum info retained.
    """
    df1 = df1.copy()
    df2 = df2.copy()
    df1['__source__'] = 'df1'
    df2['__source__'] = 'df2'

    combined = pd.concat([df1, df2], axis=0, ignore_index=True, sort=False)

    # Order rows so the priority dataset comes first within each group
    order = ['df1', 'df2'] if priority == 'df1' else ['df2', 'df1']
    combined['__source__'] = pd.Categorical(combined['__source__'], categories=order, ordered=True)
    combined = combined.sort_values(by=[key, '__source__'])

    other_cols = [c for c in combined.columns if c not in (key, '__source__')]

    # For each column, take the first non-null value within the group
    # (priority row's value wins if present, else falls back to the other row's)
    agg_dict = {c: (lambda s: s.dropna().iloc[0] if s.notna().any() else np.nan) for c in other_cols}
    result = combined.groupby(key, sort=False, as_index=False).agg(agg_dict)

    return result
```

```python
import re
import numpy as np
import pandas as pd

def convert_runtime_to_minutes(df, col='Running Time', new_col='runtime_minutes'):
    def parse_runtime(value):
        if pd.isna(value):
            return np.nan
        value = str(value)
        hours = re.search(r'(\d+)\s*hr', value)
        minutes = re.search(r'(\d+)\s*min', value)
        total = 0
        if hours:
            total += int(hours.group(1)) * 60
        if minutes:
            total += int(minutes.group(1))
        return total if (hours or minutes) else np.nan

    df[new_col] = df[col].apply(parse_runtime)
    return df

df8 = convert_runtime_to_minutes(df8, col='Running Time', new_col='runtime_minutes')
```
