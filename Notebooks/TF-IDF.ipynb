```python
import pickle
from scipy import sparse
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer

df=pd.read_csv("../../Data/New/Final_Dataset_2.csv")

# Initialize vectorizer
tfidf = TfidfVectorizer(
    stop_words='english'
)

# Fit and transform
tfidf_matrix = tfidf.fit_transform(df["tags"])

# ============================================
# Save vectorizer using pickle
# ============================================
with open("../../Data/New/tfidf_vectorizer2.pkl", "wb") as f:
    pickle.dump(tfidf, f)

# ============================================
# Save TF-IDF matrix
# ============================================
# Using scipy's sparse save (npz) instead of raw pickle here —
# tfidf_matrix is a sparse matrix and with ~1.47M rows a plain
# pickle dump would be slower and much larger on disk than npz.
sparse.save_npz("../../Data/New/tfidf_matrix2.npz", tfidf_matrix)

# ============================================
# Print results
# ============================================
print("TF-IDF matrix shape:", tfidf_matrix.shape)
print("Vocabulary size:", len(tfidf.vocabulary_))
```
